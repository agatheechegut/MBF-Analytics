```python
# 3
# fonction sigmoide


# Fonction sigmoide x=11+e-x
# Utiliser np.exp(...)
def sigmoide(x):
  return 1/(1+np.exp(-x))


# Vérification
# x=-5  0.0066928509242848554
# x= 0  0.5
# x= 5  0.9933071490757153


print("x=-5", sigmoide(-5))
print("x=0", sigmoide(0))
print("x=5", sigmoide(5))
```

```python

# 4
# échantillonnage vectorielle de Y par la fonction Sigmoide
# opération vectorielle sur série df.x . Appliquer sigmoide


df['y'] = sigmoide( df.x)


display(df)


# nombre de lignes de df. Utiliser len(…)
print("lignes de df", len(df))
```

```python
# 9
# Créer ReLU et création par iloc et loc
def relu(x):
  if x<0 :
    return 0
  else :
    return x


# vérification
# x=-5 0
# x=0 0
# x=5 5


print("x=-5", relu(-5))
print("x=0", relu(0))
print("x=5", relu(5))

```

```python
# 10
# Echantillonnage de 100 valeurs pour la fonction ReLU par répétitive


# ajout de deux colonnes xr et yr en type float
df['xr']=0.0
df['yr']=0.0


for i in range(len(df)):
  # accès à la iieme valeur de index
  t = df.index[i]
  # accès par .loc( index , nom de colonne)
  df.loc[t,'xr'] = t
  df.loc[t, 'yr'] = relu(t)


# affichage df
display(df)
```

```python
# 11
# Nombre de 0 dans la colonne yr


# Par value counts sur la 1ere valeur (la plus faible)
# utiliser .value_counts() sur la colonne y
nZeroV1 = df.yr.value_counts()[0]
# Par filtrage et taille
# utiliser len(...) df[ ... condition sur serie ...  ]
# avec la condition sur serie : df.yr==0
nZeroV2 = len( df[ df.yr ==0] )


print("nombre ZeroV1", nZeroV1)
print("nombre ZeroV2", nZeroV2)

```

```python
# 12
# Retrait des colonnes 'x' et 'y'
# Utiliser .drop( columns= [ 'nom de colonne', ...])


df= df.drop( columns= ['x','y'])


display(df)
```

```python

# 13
# tracé de yr, x direct par index
plt.plot(df.yr)
plt.grid()
plt.title('ReLU')
plt.show()
```

```python
# 14
# colonne retardé de (t-1)
df['xrB'] = df.xr.shift(1)
# colonne avancée de (t+1)
df['xrA'] = df.xr.shift(-1)


# comment est calculé xrB(t)  et xrA(t) par rapport à xr(t) ?


display(df)
```

```python
# 15
# Créer la colonne yrB type float
df['yrB'] = 0.0


# numéro de la colonne 'xrBbis'
# utiliser .get_loc('nom de colonne')
numCol2 = df.columns.get_loc('yrB')
numCol1 = df.columns.get_loc('yr')


# valeur NaN en 0
# Utiliser pd.NA et .iloc[ n°ligne, n°colonne]
df.iloc[0,numCol2] = 0
for i in range(1 , len(df)):
  df.iloc[i, numCol2] = df.iloc[ i-1 , numCol1]


display(df)
```

```python
# 23# Ajout à df colonne 'bas' de type integer
df['bas'] = 0

# Remplir la colonne par 0 : point bas V(t-1)>V(t) et V(t+1)>V(t)
# utiliser .loc[ index, 'nom colonne' ]
for t in df.index:
    if df.shift(1).loc[t, 'AAPL'] > df.loc[t, 'AAPL'] and df.shift(-1).loc[t, 'AAPL'] > df.loc[t, 'AAPL']:
        df.loc[t, 'bas'] = 1
    else:
        df.loc[t, 'bas'] = 0

display(df)
```
